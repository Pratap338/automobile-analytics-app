```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\PRATAP\Downloads\automobile_dataset\automobile_dataset.csv')
```

```python
text_columns = ['Transmission', 'Service_History', 'Color', 'Location']
```

```python
df[text_columns] = df[text_columns].fillna('Unknown')
```

```python
numeric_columns = ['Engine_Size', 'Horsepower', 'Torque', 'Accident_History', 'Fuel_Efficiency']
```

```python
for col in numeric_columns:
    middle_number = df[col].median()
    df[col] = df[col].fillna(middle_number)
```

```python
df.to_csv("cleaned_automobile_data.csv", index=False)
print("Data cleaning complete! All missing details are filled.")
```
