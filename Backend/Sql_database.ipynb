```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv('cleaned_automobile_dataset_2.csv')
```

```python
connection = sqlite3.connect('AutoMarket.db')
print("Database file created and connected successfully!")
```

```python
df.to_sql(name='vehicles', con=connection, if_exists='replace', index=False)
```

```python
sql_query = """
SELECT Make, Model, Selling_Price 
FROM vehicles 
ORDER BY Selling_Price DESC 
LIMIT 5;
"""
```

```python
results = pd.read_sql(sql_query, connection)
```

```python
print("--- Top 5 Most Expensive Cars ---")
print(results)
```

```python
connection.close()
```
