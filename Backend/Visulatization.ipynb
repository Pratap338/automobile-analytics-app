```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
connection = sqlite3.connect('AutoMarket.db')
df = pd.read_sql("SELECT * FROM vehicles", connection)
connection.close()
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))
# CHART 1: Price vs. Mileage (Scatter Plot)
sns.scatterplot(
    data=df[(df['Selling_Price'] < 100000) & (df['Mileage'] < 150000)], 
    x='Mileage', 
    y='Selling_Price', 
    alpha=0.5, 
    color='blue'
)
plt.title('Vehicle Depreciation: Price vs. Mileage', fontsize=16, fontweight='bold')
plt.xlabel('Mileage', fontsize=12)
plt.ylabel('Selling Price ($)', fontsize=12)
plt.show()
```

```python
# CHART 2: Average Price by Top 10 Brands (Bar Chart)
plt.figure(figsize=(12, 6))
# Calculate average price per brand and grab the top 10
top_brands = df.groupby('Make')['Selling_Price'].mean().sort_values(ascending=False).head(10).reset_index()

sns.barplot(data=top_brands, x='Selling_Price', y='Make', palette='viridis')
plt.title('Top 10 Most Expensive Car Brands (Average Price)', fontsize=16, fontweight='bold')
plt.xlabel('Average Selling Price ($)', fontsize=12)
plt.ylabel('Car Brand', fontsize=12)
plt.show()
```
