```python
import pandas as pd
import requests
import time
```

```python
API_KEY = 'YOUR_API_KEY_HERE'
```

```python
def ask_api_for_transmission(make, model, year):
    url = f'https://api.api-ninjas.com/v1/cars?limit=1&make={make}&model={model}&year={year}'
    try:
        response = requests.get(url, headers={'X-Api-Key': API_KEY})
        
        if response.status_code == 200:
            data = response.json()
            if len(data) > 0:
                trans_code = data[0].get('transmission', '')
                if trans_code == 'a':
                    return 'Automatic'
                elif trans_code == 'm':
                    return 'Manual'
        return 'Unknown' 
    except:
        return 'Unknown'
```

```python
df = pd.read_csv("cleaned_automobile_data.csv")
```

```python
unknown_cars = df[df['Transmission'] == 'Unknown']
print(f"Found {len(unknown_cars)} cars that need transmission data.")
```

```python
test_limit = 10
count = 0

print("Starting the API search...\n")

for index, row in unknown_cars.iterrows():
    if count >= test_limit:
        break
        
    car_make = row['Make']
    car_model = row['Model']
    car_year = row['Year']
    
    #print(f"Searching internet for: {car_year} {car_make} {car_model}...")
```

```python
real_transmission = ask_api_for_transmission(car_make, car_model, car_year)
    
    if real_transmission != 'Unknown':
        print(f"   -> Success! Found: {real_transmission}")
        df.at[index, 'Transmission'] = real_transmission
    else:
        print("   -> Not found in online database.")
        
    count += 1
    
    time.sleep(0.5)
```

```python
df.to_csv("final_enriched_automobile_data.csv", index=False)
```
